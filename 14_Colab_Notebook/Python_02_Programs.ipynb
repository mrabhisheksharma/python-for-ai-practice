## 1. Concurrency (Threading)

```python
import threading, time

def handle_client(name, wait_time):
    print(f"Serving {name} (waiting {wait_time}s)...")
    time.sleep(wait_time)
    print(f"Finished {name}!")

requests = [("Client 1", 3), ("Client 2", 1), ("Client 3", 1)]
threads = []

for name, wait in requests:
    t = threading.Thread(target=handle_client, args=(name, wait))
    threads.append(t)
    t.start()

for t in threads: t.join()
```

## 2. HTTP Fundamentals (Requests)

```python
import requests
import json

# GET Request
url = "https://api.open-meteo.com/v1/forecast?latitude=52.52&longitude=13.41&current_weather=true"
response = requests.get(url)

if response.status_code == 200:
    data = response.json()
    temp = data["current_weather"]["temperature"]
    print(f"Current Temperature in Berlin: {temp}°C")
else:
    print("Failed to fetch data")
```

```python
# POST Request
url = "https://jsonplaceholder.typicode.com/posts"
data_to_send = {
    "title": "Learning Python Requests",
    "body": "POST requests are used to send data to a server.",
    "userId": 1
}

response = requests.post(url, json=data_to_send)
print(f"Status Code: {response.status_code}")
if response.status_code == 201:
    print("Server Response:", json.dumps(response.json(), indent=2))
```

## 3. Robust APIs (Retries & Error Handling)

```python
import requests
import time

url = "https://api.open-meteo.com/v1/forecast?latitude=52.52&longitude=13.41&current_weather=true"
max_retries = 3

for attempt in range(max_retries):
    try:
        print(f"Attempt {attempt + 1}...")
        # Add timeout to prevent hanging forever
        response = requests.get(url, timeout=5)
        response.raise_for_status() # Raises Exception if status != 200
        
        temp = response.json()["current_weather"]["temperature"]
        print(f"Success! Temperature: {temp}°C")
        break # Exit loop on success
        
    except requests.exceptions.RequestException as e:
        print(f"Request failed: {e}")
        if attempt < max_retries - 1:
            print("Waiting before retrying...")
            time.sleep(2)
        else:
            print("All retries failed. Please try again later.")
```
